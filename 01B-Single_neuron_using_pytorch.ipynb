```python

import torch

# 1. Inputs
x = torch.tensor([2.0, 3.0])

# 2. Weights
w = torch.tensor([0.5, 0.2])

# 3. Bias
b = torch.tensor(0.1)

# 4. Calculate weighted sum
z = torch.dot(x, w) + b

# 5. Apply ReLU
output = torch.relu(z)

# 6. Display results
print("Weighted Sum:", z.item())
print("Neuron Output:", output.item())
```
