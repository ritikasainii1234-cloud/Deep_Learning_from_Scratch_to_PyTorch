```python


# Initial weight
w = 0.0

# Learning rate
lr = 0.1

# Repeat 10 times
for i in range(10):

    # Calculate gradient
    gradient = 2 * (w - 3)

    # Update weight
    w = w - lr * gradient

    # Calculate loss
    loss = (w - 3) ** 2

    print("Weight:", w, "Loss:", loss)

print("Final Weight:", w)
```
