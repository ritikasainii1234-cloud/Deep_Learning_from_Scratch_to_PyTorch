```python

# BACKPROPAGATION FROM SCRATCH

# Input and actual output
x = 2.0
y = 10.0

# Initial weight and bias
w = 1.0
b = 0.0


import torch

# Input and actual output
x = torch.tensor(2.0)
y = torch.tensor(10.0)

# Parameters
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# Learning rate
lr = 0.01

# Forward propagation
y_pred = w * x + b

# Loss calculation
loss = (y_pred - y) ** 2

# Backpropagation
loss.backward()

# Update parameters
with torch.no_grad():
    w -= lr * w.grad
    b -= lr * b.grad

print("Prediction:", y_pred.item())
print("Loss:", loss.item())
print("Weight:", w.item())
print("Bias:", b.item())



```
