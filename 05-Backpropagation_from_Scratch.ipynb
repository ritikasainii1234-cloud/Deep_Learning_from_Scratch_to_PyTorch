```python

# BACKPROPAGATION FROM SCRATCH

# Input and actual output
x = 2.0
y = 10.0

# Initial weight and bias
w = 1.0
b = 0.0

# Learning rate
lr = 0.01

# 1. Forward propagation
z = w * x + b
y_pred = z

# 2. Calculate loss
loss = (y_pred - y) ** 2

# 3. Calculate gradients using chain rule
dw = 2 * (y_pred - y) * x
db = 2 * (y_pred - y)

# 4. Update weight and bias
w = w - lr * dw
b = b - lr * db

print("Prediction:", y_pred)
print("Loss:", loss)
print("Weight:", w)
print("Bias:", b)


```
