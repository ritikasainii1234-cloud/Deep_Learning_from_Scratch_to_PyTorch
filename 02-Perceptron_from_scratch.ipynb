```python

import numpy as np

# 1. Dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

# 2. Initialize parameters
weights = np.zeros(2)
bias = 0.0

# 3. Hyperparameters
learning_rate = 0.1
epochs = 20

# 4. Activation function
def activation(z):
    return 1 if z > 0 else 0

# 5. Training
for epoch in range(epochs):

    for i in range(len(X)):

        # Forward propagation
        z = np.dot(X[i], weights) + bias

        # Prediction
        y_pred = activation(z)

        # Calculate error
        error = y[i] - y_pred

        # Update weights
        weights = weights + learning_rate * error * X[i]

        # Update bias
        bias = bias + learning_rate * error

# 6. Testing
print("Final Predictions:")

for i in range(len(X)):

    z = np.dot(X[i], weights) + bias

    prediction = activation(z)

    print(f"Input: {X[i]} -> Prediction: {prediction}")

# 7. Display parameters
print("Final Weights:", weights)
print("Final Bias:", bias)
```
