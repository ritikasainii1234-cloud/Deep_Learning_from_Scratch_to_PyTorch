```python

# FORWARD PROPAGATION FROM SCRATCH

# Inputs
x1 = 2
x2 = 3

# Hidden layer weights
w1 = 0.5
w2 = 0.2
w3 = 0.4
w4 = 0.3

# Hidden layer biases
b1 = 0.1
b2 = 0.2

# Hidden layer calculations
h1 = x1*w1 + x2*w2 + b1
h2 = x1*w3 + x2*w4 + b2

# ReLU activation
h1 = max(0, h1)
h2 = max(0, h2)

# Output layer weights
w5 = 0.6
w6 = 0.7

# Output bias
b3 = 0.1

# Output layer
output = h1*w5 + h2*w6 + b3

print("Hidden Neuron 1:", h1)
print("Hidden Neuron 2:", h2)
print("Final Output:", output)
```
